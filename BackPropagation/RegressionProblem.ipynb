# **Practicing and trying Backpropagation with a code example on regression small dummy datasets**

```python
import numpy as np
import pandas as pd
```

```python
data = pd.DataFrame({
    "CGPA": [2.8, 3.7, 3.1, 3.9, 3.3],
    "Resume_Score": [7.5, 5.5, 8.5, 6.5, 9.0],
    "Salary_LPA": [4.2, 6.8, 5.7, 8.5, 7.1]
})

data
```

```python
def initialize_params(layer_dims):

    np.random.seed(3)
    params = {}
    L = len(layer_dims)

    for l in range(1, L):

      params['W' + str(l)] = np.ones( (layer_dims[l-1], layer_dims[l]) ) * 0.1
      params['b' + str(l)] = np.zeros( (layer_dims[l], 1) )

    return params
```

```python
initialize_params([2,2,1]) #  1 input layer with 2 nodes for 2 features
                           #  1 hidden layer with 2 nodes
                           #  1 output layer with 1 node
```

```python
def linear_forward(A_prev, w, b):

  z = np.dot(w.T, A_prev) + b

  return z
```

```python
def L_layer_forward(X, params):

  A = X
  L = len(params) // 2

  for l in range(1, L+1):

    A_prev = A
    w = params['W' + str(l)]
    b = params['b' + str(l)]

    A = linear_forward(A_prev, w, b)

  return A, A_prev
```

```python
# Function to update the parameters
def update_params(parameters, learning_rate, y, y_hat, A1, X ):

  parameters['W2'][0][0] -= (learning_rate * (-2 * (y - y_hat)*A1[0][0]))
  parameters['W2'][1][0] -= (learning_rate * (-2 * (y - y_hat)*A1[1][0]))
  parameters['b2'][0][0] -= (learning_rate * (-2 * (y - y_hat)))

  parameters['W1'][0][0] -= (learning_rate * (-2 * (y - y_hat)*parameters['W2'][0][0]*X[0][0]))
  parameters['W1'][0][1] -= (learning_rate * (-2 * (y - y_hat)*parameters['W2'][0][0]*X[1][0]))
  parameters['b1'][0][0] -= (learning_rate * (-2 * (y - y_hat)*parameters['W2'][0][0]))

  parameters['W1'][1][0] -= (learning_rate * (-2 * (y - y_hat)*parameters['W2'][1][0]*X[0][0]))
  parameters['W1'][1][1] -= (learning_rate * (-2 * (y - y_hat)*parameters['W2'][1][0]*X[1][0]))
  parameters['b1'][1][0] -= (learning_rate * (-2 * (y - y_hat)*parameters['W2'][1][0]))
```

```python
data
```

```python
# Training Parameters

parameters = initialize_params([2,2,1])
epochs = 250

for i in range(epochs) :

  Loss = []

  for j in range(data.shape[0]) :

    X = data[['CGPA', 'Resume_Score']].values[j].reshape(2,1)
    y = data[['Salary_LPA']].values[j][0]

    y_hat, A1 = L_layer_forward(X, parameters)
    y_hat = y_hat[0][0]

    update_params(parameters, 0.001, y, y_hat, A1, X)

    loss = (y-y_hat)**2
    Loss.append(loss)

  print(f"Epoch-- {i + 1}  Loss-- {np.array(Loss).mean()}")
```

# **Let's see how backpropagation and weight updation happen by tensorflow keras**

```python
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
from tensorflow import keras

model = Sequential()
model.add(Dense(2, activation='linear', input_dim=2))
model.add(Dense(1, activation='linear'))
```

**Setting the initial weights as our was**

```python
model.get_weights()
```

```python
new_weights = [np.array([[0.1 , 0.1],
        [0.1 , 0.1 ]], dtype=np.float32),
 np.array([0., 0.], dtype=np.float32),
 np.array([[ 0.1 ],
        [0.1]], dtype=np.float32),
 np.array([0.], dtype=np.float32)]
```

```python
model.set_weights(new_weights)
```

```python
model.get_weights()
```

```python
optimizer = keras.optimizers.Adam(learning_rate=0.001)
model.compile(loss='mean_squared_error', optimizer=optimizer)
```

```python
model.fit(data[['CGPA', 'Resume_Score']], data['Salary_LPA'], epochs = 250, verbose = 1, batch_size=1) #Keras took more epochs to converge
```
