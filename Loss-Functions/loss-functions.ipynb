# **Some important classification and regression loss funtions**

# **Regression Loss Functions**

```python
import numpy as np 

# Dummy y and y_pred

y = np.array([16.7,18.6,13.1,19.2,17.3])
y_pred = np.array([16.2,17.9,12.6,20.7,17.7])
```

```python
# MSE

def Mean_Squared_Error(y, y_pred):
    
    n = len(y)
    error = np.sum(np.square(y-y_pred)) / n
    
    return error

mse = Mean_Squared_Error(y, y_pred)
mse
```

```python
# MAE

def Mean_Absolute_Error(y, y_pred):
    
    n = len(y)
    error = np.sum(np.abs(y-y_pred)) / n
    
    return error

mse = Mean_Absolute_Error(y, y_pred)
mse
```

```python
# Huber Loss 

def Huber_Loss(y, y_pred, delta=1.0):
    
  error = y - y_pred
  is_small_error = np.abs(error) <= delta

  squared_loss = 0.5 * np.square(error)
  linear_loss = delta * (np.abs(error) - 0.5 * delta)

  return np.mean(np.where(is_small_error, squared_loss, linear_loss))

huber_loss = Huber_Loss(y,y_pred, delta = 0.5)
huber_loss
```

# **Classification**

**Binary**

```python
# Dummy y and pred_y

y = np.array([1, 0, 1, 1, 0])

# predicted probabilities (output of a Sigmoid function, between 0 and 1)
y_pred_binary = np.array([0.9, 0.1, 0.8, 0.6, 0.3])
```

```python
# Binary-Cross Entropy / Log_Loss

def Log_Loss(y, y_pred):

    eps = 1e-16
    y_pred = np.clip(y_pred, eps, 1 - eps)

    loss = -(np.mean((y * np.log(y_pred)) + ((1- y)* np.log(1-y_pred))))

    return loss 

logloss = Log_Loss(y, y_pred)
print(f"Log Loss: {logloss:.2f}")
```

```python
# Categorical-cross entropy

y = np.array([[1, 0, 0], [0, 1, 0], [0, 0, 1], [1, 0, 0], [0, 1, 0]])

# Predicted probabs using softmax
y_pred = np.array(
    [[0.85, 0.10, 0.05], [0.15, 0.75, 0.10], [0.05, 0.15, 0.80], [0.60, 0.30, 0.10], [0.20, 0.50, 0.30]]
)
```

```python
def Categorical_Cross_Entropy(y, y_pred):
    
  eps = 1e-15
  y_pred = np.clip(y_pred, eps, 1 - eps)
  loss = -np.mean(np.sum(y * np.log(y_pred), axis=1))

  return loss


cce_loss = Categorical_Cross_Entropy(y, y_pred)
cce_loss
```

```python
# Hinge Loss

y = np.array([1, -1, 1, 1, -1])

# prediction/ scores (logits, not probabilities)
y_pred = np.array(
    [1.5, -0.8, 0.4, 2.1, -1.2]
) 


def Hinge_Loss(y, y_pred):
  # Formula: max(0, 1 - y * y_pred)
  losses = np.maximum(0, 1 - y * y_pred)
  return np.mean(losses)


hingeloss = Hinge_Loss(y, y_pred)
hingeloss
```
